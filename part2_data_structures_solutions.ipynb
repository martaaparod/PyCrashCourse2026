## Core Data Structures

The 4 most common ways to store data in Python are:
1. Lists: ordered collection of data, mutable (we can change, add or remove items after the list has been created)
2. Tuples: ordered collection of data, immutable (we cannot change, add or remove items after the tuple has been created)
3. Dicts: collection of key-value pairs
4. Sets: unordered collection of unique items (duplicate values are automatically removed)

Let's start using lists:

```python
my_list = [1, 2, 3, 4, 5]
```

To access an element in the list, we use its index. In Python indices always start at 0. You can also access indices from the end of the list.

| Element        | 1st | 2nd | ... | 2nd to last | Last |
|----------------|-----|-----|-----|-------------|------|
| Positive index | `0` | `1` | ... | `n-2`       | `n-1`|
| Negative index | `-n`| `-(n-1)` | ... | `-2`   | `-1` |

Here `n` is the length of the list.

```python
print(len(my_list))
```

```python
print(my_list[0])
print(my_list[-2])
```

To access multiple elements of a list, we can use slicing, defined using `:`. The number before the colon is the starting index (default `0`), and the number after is the ending index, which is exclusive (default is the end of the list). For example:

```python
print(my_list)
print(my_list[1:3])  # elements at index 1 and 2
print(my_list[:3])   # from the start up to index 2
print(my_list[2:])   # from index 2 to the end
print(my_list[:])    # the entire list
```

Lists can also contain elements of different types.

```python
my_list[1] = 100
my_list[2] = 'hello'
print(my_list)
```

```python
my_list[5] = 3
```

Methods are functions that belong to a specific object and are called using dot notation: `my_list.append(3)`. Lists have several built-in methods for common operations.
- `append(x)`: adds an element to the end
- `pop(i)`: removes and returns an element by index (defaults to last)
- `sort()`: sorts the list in place

```python
my_list.pop(2)
```

```python
print(my_list)
```

As well as methods, Python provides several built-in functions that are useful for working with lists:

```python
print(len(my_list))       # number of elements
print(sum(my_list))       # sum of all elements
print(min(my_list))       # smallest element
print(max(my_list))       # largest element
print(sorted(my_list))    # returns a new sorted list
```

If we try to do similar operations with tuples, we will get an error:

```python
my_tuple = (1, 2, 3)
my_tuple[0] = 10
```

```python
# dictionaries
person = {"name": "Alice", "age": 25}
print(person["name"])
person["profession"] = "teacher"
```

```python
print(person["Bob"])
```

You can also iterate through the keys in a dictionary using `.keys`:

```python
for key in person.keys():
    print(f"Key: {key}, Value: {person[key]}")
```

```python
# sets
fruits = {"apple", "banana", "apple"}
print(fruits)  # duplicates removed
```

## NumPy

In Python, you use the `import` keyword to make code in one module available in another. You may want to import your own code, for example a library of functions you want to access (without re-defining everything), or a class (an object-contructor script). This is really helpful for organizing your code.

You can also import modules and packages available online (after download and installation). Each module/package has online documentation explaining the usage of each function/class.

NumPy is one of the most commonly used packages in Python. The documentation can be found in https://numpy.org/devdocs/index.html. Let's begin by importing the package!

```python
import numpy as np
```

Python lists are flexible, but they were not designed for numerical computation. Consider adding two vectors:

```python
a = [1, 2, 3]
b = [4, 5, 6]
a + b
```

NumPy arrays behave the way we expect:

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
a + b
```

A NumPy array has three key properties: `dtype` (the data type of its elements - all elements must have the same data type!), `shape` (its dimensions as a tuple), and `ndim` (the number of dimensions).

```python
matrix = np.array([[1, 3, 5], [2, 4, 6]])

print(matrix.dtype)
print(matrix.shape)
print(matrix.ndim)
```

There are different ways of creating arrays.

```python
# arrays of zeros and ones
print(np.zeros(5))
print(np.ones((2, 3))) # 2 rows, 3 columns
```

```python
# evenly spaced values
print(np.arange(10))
print(np.arange(start=0, stop=10, step=2))
dt = 0.1
print(np.arange(start=2, stop=10, step=dt))

# if you want to specify the number of equally-spaced values between start & stop:
print(np.linspace(start=0, stop=1, num=5))
```

NumPy arrays support the same slicing syntax as Python lists, and extend it naturally to multiple dimensions: `a[0, :]` selects the first row, `a[:, 1]` selects the second column.

### Vector operations

Arithmetic operators (`+`, `-`, `*`, `/`, `**`) apply element-wise to NumPy arrays. NumPy also provides additional mathematical functions via `np.`, such as `np.sqrt()`, `np.sin()`, and `np.dot()`. For a full list, see the NumPy function reference.


```python
a = np.array([1, 4, 9])
b = np.array([2, 3, 4])

print(a + b)
print(a * 2)
print(np.sqrt(a))
print(np.dot(a, b))
```

### Concatenation

At times you may still want to combine or concatenate vectors. You can do this through `np.concatenate`. By default, concatenation occurs acros the first dimension (`0`), but you can also specify the dimension.

```python
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6],
              [7, 8]])

print(np.concatenate([a, b]))
print(np.concatenate([a, b], axis=1))
```

For concatenating along the `0` and `1` dimension, you can simply use `np.vstack` and `np.hstack`.

```python
print(np.vstack([a, b]))
print(np.hstack([a, b]))
```

### Broadcasting

When an operation involves arrays of different shapes, NumPy automatically expands the smaller array to match. This is called broadcasting.

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
a + 10        # adds 10 to every element
```

Note that dimensions must be compatible for broadcasting to work.

```python
a + np.arange(4)
```

### Linear Algebra

NumPy provides tools for the most common linear algebra operations you will encounter. A common point of confusion is the difference between element-wise multiplication and matrix multiplication. By default, arithmetic operators act element-wise. To perform matrix multiplication, use the `@` operator:

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print(A * B)
print(A @ B)
```

We can use `np.linalg` to perform matrix operations such as computing the inverse, determinant or the eigenvelues and eigenvectors of a matrix.

```python
np.linalg.det(A)
```

```python
np.linalg.inv(A)
```

```python
np.linalg.eig(A)
```

To solve a linear system of the form $Ax = b$, we can compute $x = A^{-1}b$.

```python
x =  np.linalg.inv(A) @ b
print(x)
```

However, `np.linalg.solve` is preferred over computing the inverse explicitly because it is faster and more stable numerically.

```python
x_solve = np.linalg.solve(A, b)
print(x_solve)
```

We can also compute the norm of a vector using `np.linalg.norm`.

```python
print(np.linalg.norm(A@x - b))
print(np.linalg.norm(A@x_solve - b))
```

The residual is not zero because of floating point arithmetic. Computers cannot represent real numbers exactly, so small rounding errors accumulate. This value is effectively zero.

### Random numbers

We can also use NumPy to sample from a range of distributions. Here are a few examples:

```python
A = np.random.uniform(size=(3,3)) # you can specify the range here with low= and high=
b = np.random.rand(3) # range is by default [0, 1]
print(A)
print(b)
```

Let's compute some statistics:

```python
print("Mean:", A.mean())
print("Variance:", A.var())
print("Standard Deviation:", A.std())
```

For more advanced mathematical operations (integration, solving differential equations, optimisation) the SciPy library builds directly on NumPy and is worth exploring. See the SciPy documentation https://docs.scipy.org/doc/scipy/ for a full overview.
