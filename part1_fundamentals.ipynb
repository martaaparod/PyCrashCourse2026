## Variables and Types

A variable is a named container that stores a value. You create one using the assignment operator `=`. The name goes on the left, the value on the right. Python remembers the value so you can use it later.

```python
x = 5
name = "Alice"
```

The most common built-in types you will encounter are:

- `int`: whole numbers, `3`, `-7`
- `float`: decimal numbers, `3.14`, `-0.001`
- `str`: text, `"hello"`, `"x = 5"`
- `bool`: logical values, `True` or `False`

We will meet further types (lists, dictionaries and others) in Session 2.

Python is dynamically typed: it automatically determines the type of a variable from the value assigned to it.

```python
a = 1
print(type(a))
```

```python
b = 1.0
print(type(b))
```

```python
c = "1"
print(type(c))
```

You can explicitly convert a variable from one type to another. This is called type casting:

```python
w = bool(0)      # int → bool
x = int(3.14)    # float → int
y = float(5)     # int → float
z = str(42)      # int → str

print(w, x, y, z)
```

**CAUTION**: It doesn't always work!

```python
x = int("hello")
```

```python
y = int(3.9)
print(y)
```

## Control Flow

Control flow refers to the order in which Python executes your code. By default, Python runs line by line from top to bottom but control flow structures let you change that default behaviour in two fundamental ways:

1. **Conditionals**

`if`/`elif`/`else` allow you to execute a block of code only if a condition is satisfied.

```python
x = 1
if x > 0:
    print("positive")
elif x == 0:
    print("zero")
else:
    print("negative")
```

2. **Loops**

`for`/`while` loops allow you to repeat a block of code several times.

```python
# repeats 5 times
for i in range(5):
    print(i)
print("Finished for loop!")

# repeats until x hits 0
while x > 0:
    x -= 1
    print(x)
print("Finished while loop!")
```

**CAUTION**: A `while` loop will run forever if its condition is never met! If this happens, your notebook will appear to hang. You can stop it by clicking 'Kernel', 'Interrupt' in Jupyter. Always check that your loop has a clear path to termination before running it.



```python
x = 1
while x > 0:
    x += 1
```

`for` loops don't have this problem. Make sure to choose the appropriate loop for the task you perform.

## Functions

A function is a block of code which only runs when it is called. It is defined using the `def` keyword, followed by a name and parentheses.

```python
def print_hello():
    print("Hello! This is a function!")
```

```python
print_hello()
print_hello()
print_hello()
```

Functions take as input arguments and can optionally return an output.

```python
def quadratic_function(a, b, c, x):
    f = a * x**2 + b * x + c
    return f
```

```python
for y in range(3):
    val = quadratic_function(1, 2, 3, x=y)
    print(val)
```
