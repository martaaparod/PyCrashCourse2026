## Readable Code

Let's begin by looking at what you shouldn't do. Try figuring out what these functions do!

```python
def function1(f, g, x0, e=1e-6, n=100):
    x = x0
    for i in range(n):
        z = f(x)
        if abs(z) < e:
            return x
        x = x - z / g(x)
    raise ValueError("Did not converge")
```

```python
def function2(f, a, b, n):
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    y = np.array([f(xi) for xi in x])
    s = y[0]/2 + np.sum(y[1:-1]) + y[-1]/2
    return h * s
```

```python
def function3(f, df, p, a=0.01, e=1e-6, n=1000):
    q = p
    for i in range(n):
        d = df(q)
        if abs(d) < e:
            return q
        q = q - a * d
    raise ValueError("Did not converge")
```

```python
def function4(d, t=0.5):
    r = []
    for i in range(len(d)):
        s = 0
        for j in range(len(d[i])):
            s += d[i][j]
        a = s / len(d[i])
        if a > t:
            r.append(d[i])
    m = 0
    for i in range(len(r)):
        s = 0
        for j in range(len(r[i])):
            s += r[i][j]
        m += s / len(r[i])
    if len(r) > 0:
        m = m / len(r)
    return r, m
```

If you had to rewrite these functions, what would you do differently?

The standard format for writing Python code is PEP 8 (https://peps.python.org/pep-0008/), which covers naming conventions, indentation, spacing, etc. Most modern code editors and IDEs (such as PyCharm, VS Code, Jupyter) will automatically flag PEP 8 violations and suggest corrections as you type.

### Naming conventions:
- Variables: lowercase with underscores, `learning_rate`, `max_iterations`
- Functions: lowercase with underscores, named as a phrase describing what the function does `compute_gradient()`, `is_prime()`
- Global constants (usually it is better to use default arguments in functions but you might encounter these): uppercase with underscores, `MAX_ITER = 100`, `TOLERANCE = 1e-6`
- Classes (not covered in this course): capitalised words with no underscores, `LinearRegression`, `DataLoader`

Try to avoid names that are only one letter (except established notations such as `x`, `y`, `A`) and abbreviations that are not universally known (if you are not sure, it is safer to use a longer name). As you saw in the exercise above, they make code very hard to read.

### Comments and docstrings:
- Comments are written preceded by the `#` and explain why the code does something. A good comment adds context that cannot be inferred from the code alone. Note that a well-named function and clear variable names often make comments unnecessary, so make sure you are aware of why you are adding a comment and what it adds.
- Docstrings are strings written between `"""` symbols at the top of a function. They describe what the function does, and what its inputs and outputs are. For simple functions, a single line is enough to explain, e.g., `"""Returns the area of a circle with radius r"""`. Not every function needs the full format, only ones with multiple parameters and non-obvious behaviour.

### Function structure:
- A function should do one thing (and do it clearly). If you find yourself writing a function that computes, formats, and saves results all in one block, split it into three. Smaller functions are usually easier to read and reuse.
- Avoid hard-coded values with no explanation of what it represents or where it comes from. Instead, replace them with named constants or default arguments.

## Debugging

Debugging is the process of identifying and fixing errors in your code. You will often spend a significant amount of time debugging, sometimes more than writing code in the first place.

```python
def compute_mean(values):
    return sum(values) / len(values)

compute_mean([])
```

When reading a traceback error, keep in mind the following:
- The last line is the actual error, the lines above show the call stack leading to it
- Check the error type (ZeroDivisionError, TypeError, IndexError, etc.), as it will tell you something specific about what went wrong
- The line number is your starting point

The simplest debugging strategy is to insert `print()` statements at key points in your code to inspect the values of variables as the code runs. This will allow you to confirm if the logic in every line leading up to the printed statement is correct or not, and narrow down where the mistake is.

**Exercise:** find the mistake in this code

```python
def matrix_vector_multiply(A, x):
    """Multiply matrix A by vector x without using @"""
    n = A.shape[0]
    result = np.zeros(n)
    for i in range(n):
        for j in range(n):
            result[i] += A[i, j] * x[j]
    return result
```

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
x = np.array([1, 1, 1])

print(matrix_vector_multiply(A, x))
```

**Note:** once you have identified and fixed the bug, don't forget to remove the print statements in your final version of the code!

You can also check correct statements using `assert`:

```python
assert abs(compute_mean([1, 2, 3]) - 2.0) < 1e-10
assert abs(compute_mean([1, 2, 3]) - 0.0) < 1e-10
```

Most professional code editors include a built-in debugger. This is a tool that lets you pause code execution at any line and inspect the state of all variables interactively, without adding and removing print statements. PyCharm and VS Code both have built-in debuggers: you set a breakpoint by clicking next to a line number, run the code in debug mode, and execution pauses at that point so you can step through the code line by line and watch variables change in real time. Check out online tutorials on how to use this for your choice of editor.
